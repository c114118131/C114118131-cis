```python
# 整數：班級學生人數
student_count = 60

# 浮點數：平均每晚睡眠時數（小時）
average_sleep_hours = 6.7

# 字串：課程名稱
course_name = "Python與資料科學"

# 布林值：是否有缺失值（True/False）
has_missing_value = True

# 依序印出每個變數的值與型態
print("student_count =", student_count, "型態:", type(student_count))
print("average_sleep_hours =", average_sleep_hours, "型態:", type(average_sleep_hours))
print("course_name =", course_name, "型態:", type(course_name))
print("has_missing_value =", has_missing_value, "型態:", type(has_missing_value))
```
